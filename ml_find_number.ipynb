**description:**

KNN linear -
draw y=ax+b line and find the closest dot from a item you are looking for to get forcast result.

logistic - a statistical model used for binary regression. Binary regression is used when data has 2 answers. pass/fail the test, cat/dog.

polynomial - add all plot in xy graph and if plot looks like curbed line, use this regression. Ex. salary wth years of experience

support vector - use this regression when plot needs to find averages or each x point and create a traight line.

visualization decision tree - use this regression when plot can be separated by some groups.
